```python
import json
with open('data/data_test.json', 'r') as f:
    data_test = json.load(f)
```

```python
import pandas as pd
data_test = pd.DataFrame(data_test)
```

```python
from tqdm import tqdm
import numpy as np
answers = []
for i in tqdm(data_test.text):
    answers.append(np.argmax(model(tokenizer(i, return_tensors='pt',
                                            truncation=True, max_length=512
                                             ).to('cuda')['input_ids']).logits.detach().to('cpu'), axis=1).item())
```

```python
len(answers)
```

```python
data_test['predict'] = answers
```

```python
labels_dict = {
    0: "human_text",
    1: "mixed_text",
    2: "machine_text"
}
```

```python
data_test.predict = data_test.predict.apply(lambda x: labels_dict[x])
```

```python
from sklearn.metrics import classification_report

print(classification_report(data_test.label, data_test.predict))
```
